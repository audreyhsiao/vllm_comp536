```python
import json

def load_qwen_trace(path, workload_type="online_trace_A"):
    records = []
    with open(path, "r") as f:
        for line in f:
            rec = json.loads(line)
            records.append({
                "workload_type": workload_type,
                "chat_id": rec["chat_id"],
                "parent_chat_id": rec["parent_chat_id"],
                "arrival_time": rec["timestamp"],
                "input_length": rec["input_length"],
                "output_length": rec["output_length"],
                "req_type": rec["type"],
                "turn": rec["turn"],
                "blocks": rec["hash_ids"],   # 一串 block id，block_size=16 tokens
            })
    return records

traceA = load_qwen_trace("qwen-bailian-usagetraces-anon/qwen_traceA_blksz_16.jsonl", "online_trace_A")
traceB = load_qwen_trace("qwen-bailian-usagetraces-anon/qwen_traceB_blksz_16.jsonl", "online_trace_B")
```
